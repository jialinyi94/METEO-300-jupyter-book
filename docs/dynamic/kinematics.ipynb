# Kinematics

这节 Lesson 9 整体在干一件事：

给你一套“纯几何、纯运动”的语言，用来描述大气怎么动，而暂时不问“为什么会这么动”。

也就是 kinematics（运动学）：只看风场的形状、拉伸、旋转、发散收缩，以及这些水平运动怎么通过质量守恒“逼出”垂直运动。

我按小节给你拆开讲，并把数学和物理图像连起来。

## 1. Overview：从卫星云图出发的“运动解剖学”

卫星水汽/红外云图上，你看到的是一堆卷曲的云带、漩涡和拉伸带——复杂，但并非无限复杂。

课程的核心主张是：

任何水平气流，都可以拆成 5 种基本运动类型的组合：
1.	平移（translation）
2.	伸展形变（stretching deformation）
3.	剪切形变（shearing deformation）
4.	涡度/旋转（vorticity）
5.	散度 / 辐合（divergence / convergence）

本课的学习目标是：
- 能在天气图上辨认收束/发散、正负涡度
- 能从观测计算上述几种流型的强度
- 能把垂直运动和水平散度/辐合联系起来（“哪儿在下沉、哪儿在上升”）

## 2. 9.1 流线 vs 轨迹：快照 vs 真实轨迹

流线（streamlines）
- 定义：在某一固定时刻，空间中处处与瞬时速度矢量平行的曲线。只看方向，不看速率大小；有时会画“更密代表风更强”，但不是必须。
- 你可以把它当成风场的瞬时“纹理图”，是一张 snapshot；时间一变，流线图也变。

在流线图上：
- Confluence：流线之间距离变小，线“往一起挤”；
- Diffluence：流线之间距离变大，线“往外散开”。

轨迹（trajectories）
- 空气质点随时间真实走过的路径——既包含方向也包含速度（时间信息）。
- NOAA HYSPLIT 这种模式给出的就是风迹线，图上轨迹上的小方块等时间间隔（比如每 6 小时一个），方块变密→走得慢（收敛），变疏→走得快（发散）。

非常关键的一点区分：
- Confluence / Diffluence：只和流线间距有关，是“几何收束/发散”；
- Convergence / Divergence：和速度大小沿流线方向的变化有关，是“速度收束/发散”。

简单比喻：
- 流线像“路网线条”：路越挤越密，是 confluence；路越岔越开，是 diffluence。
- Convergence 则是：同一条路上，车速前慢后快 → 车相互拉开；前快后慢 → 车挤在一起。


## 3. 9.2 看气块如何被“四块偏导数”拆解

这一节把你从“看风羽”推进到“看速度场的空间导数”。

假设一个水平气块（air parcel），我们只看二维 $(x, y)$：
- 速度场：$u(x,y)$, $v(x,y)$
- 取一个参考点 $(x_0,y_0)$，周围小位移 $dx$, $dy$, 用二维 Taylor 展开：
$$u(x_0+dx,y_0+dy) \approx u_0 + \frac{\partial u}{\partial x}dx + \frac{\partial u}{\partial y}dy$$
$$v(x_0+dx,y_0+dy) \approx v_0 + \frac{\partial v}{\partial x}dx + \frac{\partial v}{\partial y}dy$$

其中：
- $u_0, v_0$：整个气块的平移速度（translation）
- 四个偏导：
- \partial u/\partial x：x 方向上，u 随 x 的变化（“左右方向拉伸/挤压”）
- \partial v/\partial y：y 方向上，v 随 y 的变化（“南北方向拉伸/挤压”）
- \partial v/\partial x：沿 x 方向，南北速度 v 的变化（类似“斜着推”）
- \partial u/\partial y：沿 y 方向，东西速度 u 的变化（另一个“斜着推”）

教材强调一个小但重要的 sign 规则：
- 偏导 > 0：
- 正值变得更大，或负值变得不那么负（数轴上往右）。
- 偏导 < 0：
- 正值变小，或负值变得更负（数轴上往左）。

这四个偏导的组合，就会产生后面说的四种形变 + 旋转 + 散度。

## 4. 9.3 五种基本气流类型

这节是 Lesson 9 的核心：“所有水平风场 = 5 种基本动作的叠加”。
	1.	平移（Translation）
- 整块气团一起移动，形状、面积、取向都不变。
- 数学上：只出现 u_0, v_0，没有任何偏导数项。
	2.	伸展形变（Stretching deformation）
- 用一对偏导来度量（教材公式位置是占位符，大意是：
- \partial u/\partial x > 0：两边往两侧拉，x 方向变长
- \partial v/\partial y < 0：南北向收缩，y 方向变短）
- 条件是两者大小相等但符号相反时，总面积不变，只是长宽比变了。
- 结果：
- 沿某一轴“拉长”（diffluence），垂直方向“压扁”（confluence）。
- 物理意义：
- 沿一条线把冷暖空气拉开 → 产生锋面带的一种机制（frontogenesis 的一部分）。
	3.	剪切形变（Shearing deformation）
- 由 \partial v/\partial x 和 \partial u/\partial y 组合。
- 效果：方形气团被剪成菱形：
- 左下往一个方向推，右上往相反方向推。
- 同样总面积不变，但轴向发生倾斜；沿一条对角线拉长，另一条对角线压缩。
- 正号：SW–NE 方向拉长、SE–NW 收缩；负号反过来。
- 物理含义：配合伸展形变，一起塑造锋区几何结构。
小结：
伸展 & 剪切，都有“某方向 diffluence + 垂直方向 confluence”，只是拉伸轴不同。两者都和锋区的形成/削弱强相关。
	4.	涡度（Vorticity, ζ）
- 垂直分量：
\zeta = \frac{\partial v}{\partial x} - \frac{\partial u}{\partial y}
（教材对应式 [9.2]）
- 只产生纯旋转——气块不拉伸、不剪切、面积不变，只整体转动。
- 右手定则：手指顺着流线弯，拇指指向 vorticity 的方向：
- 在北半球，逆时针（cyclonic）→ ζ > 0；顺时针（anticyclonic）→ ζ < 0。
- 因此：
- 北半球低压系统：逆时针环流 → 正涡度
- 北半球高压系统：顺时针 → 负涡度
- 重要性：
- 绝大部分中纬天气系统都可以用“涡度中心 + 前后涡度发展”来描述。
	5.	散度（Divergence, δ）
- 水平散度定义：
\delta_H = \frac{\partial u}{\partial x} + \frac{\partial v}{\partial y}
（教材用希腊小写 delta 表示）
- δ > 0：气块面积增大（膨胀）→ 散度；
- δ < 0：气块面积缩小 → 辐合（convergence）。
- 严格来说，这是水平散度；再加上 \partial w/\partial z 就是三维散度。

教材最后用矢量形式写：
\delta = \nabla \cdot \mathbf{U}
并强调五种运动的单位都是 s⁻¹，是一种“频率/速率”。

5. 9.4 散度 = 面积（或体积）的分数变化率

这一节是把“散度”从抽象符号变成几何上的面积/体积变化。

设一个小矩形气块，边长 Δx, Δy，不同角上的速度略有不同。教材推导（略去中间代数）得到：

\frac{1}{A}\frac{dA}{dt} = \frac{\partial u}{\partial x} + \frac{\partial v}{\partial y} = \delta_H
- A：气块面积
- 左边：面积相对变化率（单位 s⁻¹）
- 右边：水平散度

因此：

水平散度 = “气块面积的相对增长率”

δ > 0 → 面积随时间变大（散度）
δ < 0 → 面积变小（辐合）

拓展到三维体积 V，有类似的关系：

\frac{1}{V}\frac{dV}{dt} = \nabla \cdot \mathbf{U}

所以：

三维散度 = 气块体积的相对变化率。

教材给了一个练习：
- 面积 = 10,000 km²，每秒长大 1 km² → δ ≈ 10⁻⁴ s⁻¹。
- 如果 δ = –10⁻⁴ s⁻¹ → 面积在缩小。

这也顺带告诉你一个量级感：天气尺度上的典型散度 ~10⁻⁵–10⁻⁴ s⁻¹。

6. 9.5 水平散度/辐合如何“逼出”垂直运动？

这里开始调动 质量守恒方程（完整推导在 Lesson 10）：

\frac{D\rho}{Dt} + \rho (\nabla \cdot \mathbf{U}) = 0
- ρ：密度；D/Dt 为物质导数（跟着气块走）。

若某处 散度 > 0（体积变大），质量守恒要求：要么密度变小，要么有别的方向上的补偿。

教材做了一个近似：

“在同一高度面上，密度 ρ 随时间变化可以近似看成 0（虽然随高度 z 是指数衰减，但对每个 z 面来讲，ρ 大致常数）。”

于是质量守恒简化为：

\nabla \cdot \mathbf{U} = \frac{\partial u}{\partial x}+\frac{\partial v}{\partial y}+\frac{\partial w}{\partial z} = 0

换句话说：

\underbrace{\left(\frac{\partial u}{\partial x}+\frac{\partial v}{\partial y}\right)}{\text{水平散度}}
+
\underbrace{\frac{\partial w}{\partial z}}{\text{垂直散度}}
= 0

所以：
- 水平散度 > 0 → \partial w/\partial z < 0：
- w 随高度减小：上面 w 小、下面 w 大 → 整体表现为下沉。
- 水平辐合 (δ<0) → \partial w/\partial z > 0：
- w 随高度增大：下面 w 小、上面 w 大 → 整体表现为上升。

再加上两个边界条件：
- 地面：w ≈ 0（固体边界）
- 对流层顶附近：位温急增，强烈抑制穿透，w ≈ 0。

这意味着垂直速度在对流层中间某处达到极值（上升最强/下沉最强），两端为 0。

教材用图和表总结了四种情形（“近地面 vs 高空 × 辐合/散度”）：

平面 & 过程	面积变化	∂w/∂z 符号	w 方向
近地面辐合	减小	+	向上
近地面散度	增大	–	向下
高空辐合	减小	+	向下
高空散度	增大	–	向上

天气上的经典翻译：
- 低空辐合 + 高空散度 → 上升运动 → 低压、云雨
- 低空散度 + 高空辐合 → 下沉运动 → 高压、晴朗

这就是“高空散度抽吸地面空气 → 地面气压降低”的标准故事。

7. 9.6 垂直风：有多大、多快、哪层最强？

上一节告诉你 ∂w/∂z 怎么由水平散度决定，这一节进一步把 w(z) 算出来，并给你量级直觉。

从
\frac{\partial w}{\partial z} = - \delta_H(z)
出发，积分：

w(z) - w(0) = - \int_0^z \delta_H(z’)\, dz’

假定地面 w(0) ≈ 0，则：

w(z) = - \int_0^z \delta_H(z’)\, dz’
这是 kinematic vertical velocity 的定义。

进一步的简化假设：

观测发现，大尺度（synoptic scale，~1000 km）的水平散度 δ_H 随高度大致线性变化：

\delta_H(z) = \delta_s + b z
- δ_s：地面散度
- b：常数（随具体情况定）

还知道：
- 在某个高度 z_{LND}（nondivergence level），水平散度为 0：
\delta_H(z_{LND}) = 0。

利用这些条件，可以解出 b、再代回积分，得到 w(z) 的抛物线剖面：在地面 & 对流层顶为 0，中间达到极值。教材给出的典型数值：
- δ_s ~ 10⁻⁵ s⁻¹（大尺度地面散度）
- z_{LND} ~ 5000 m

代入后：
- 垂直速度 w 的量级只有 几 cm/s。
- 一整天（~10⁵ s），空气可以升/降 几 km——这与对流层厚度同量级。

教材把这种 w(z) 的剖面形状叫做 “弓弦模型（bowstring model）”：两端固定，中间弯曲。

对比：大尺度 vs 对流尺度
- 大尺度（低压、高压系统）：w ~ cm/s
- 强雷暴上升气流：w ~ 几 m/s 甚至十几 m/s

也就是说：
- 低压、高压：慢火小炖，大范围、长期控制天气格局；
- 雷暴、积雨云：局地高强度“电梯”，短时间把气块送上对流层顶。

8. Summary：用 5 种动作 + 散度关系看懂“风的结构”

Lesson 9 的总结把整个故事串起来：
	1.	Kinematics = 描述，不问成因
- 流线 → 风场的时间快照
- 轨迹 → 单个气块的真实路径
	2.	全部水平运动 = 5 种基本类型组合
- 平移：整体移动
- 伸展 & 剪切：贡献锋区的形成/削弱（frontogenesis/frontolysis）
- 涡度：低压/高压的旋转结构
- 散度：气块膨胀 or 收缩
	3.	散度 + 质量守恒 → 垂直运动
- 高空散度 ↔ 低层辐合 ↔ 上升 ↔ 云雨、低压
- 高空辐合 ↔ 低层散度 ↔ 下沉 ↔ 晴天、高压
- 垂直风的大尺度量级 ~ cm/s；用 bowstring 模型可以近似 w(z)。
	4.	数学工具已经齐活
- 你现在能从天气图上定量计算：伸展/剪切形变、涡度、散度、以及由此导致的垂直运动。
- 下一课（Dynamics – Forces）才会引入“这些运动为什么会发生”（压力梯度力、科氏力、摩擦等），把kinematics 的几何结构和动力学的力平衡拼起来。
